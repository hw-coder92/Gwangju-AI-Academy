```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN , Dense , Input
```

```python
x = np.array([
    [1,2,3],
    [2,3,4],
    [3,4,5],
    [4,5,6],
])

y = np.array([4,5,6,7])

x = x.reshape(-1,3,1)

print(x)
print(x.shape)

# RNN 은 입력 3차원(data size, time_steps, features)
model = Sequential([
    Input(shape=(3,1)),
    SimpleRNN(2),
    Dense(1)
])

model.summary()

model.compile(
    loss = "mse",
    optimizer = "adam"
)

model.fit(
    x,
    y,
    epochs = 300,
    verbose=1
)
```

```python
# 예측
x_pred = np.array([[5,6,7]]).reshape(-1,3,1)
# pred = model.predict(np.array([[5],[6],[7]]))
pred = model.predict(x_pred)

print(pred)
```

```python
# 주식 종가 예측

x = np.array([
    [50000, 51000, 52000, 53000],
    [51000, 52000, 53000, 54000],
    [52000, 53000, 54000, 55000]
])

# 다음 날 종가
y = np.array([
    54000,
    55000,
    56000
])

x = x.reshape(-1,4,1)
print(x.shape)

# 모델 생성

# data_size = 3
# time_steps = 4
# features = 1 (종가만 사용)

model = Sequential([
    Input(shape=(4,1)),
    SimpleRNN(8, activation= "relu"),
    Dense(1)
])

model.summary()

model.compile(
    loss = "mse",
    optimizer = "adam"
)

model.fit(
    x,
    y,
    epochs = 500,
    verbose=1
)

```

```python
# 예측
# [53000, 54000, 55000, 56000]

x_pred = np.array([
    [53000, 54000, 55000, 56000]
])
x_pred = x_pred.reshape(-1,4,1)

pred = model.predict(x_pred)
print("다음 날 예상 종가",pred)
```

```python
# RNN 을 푸시오.

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, SimpleRNN, Dense

# ==========================
# 종가(Close) 데이터
# 최근 3일 종가 -> 다음날 종가
# ==========================

x = np.array([
    [[70000], [70500], [71000]],
    [[70500], [71000], [71500]],
    [[71000], [71500], [72000]],
    [[71500], [72000], [72500]],
    [[72000], [72500], [73000]]
], dtype=np.float32)

y = np.array([
    71500,
    72000,
    72500,
    73000,
    73500
], dtype=np.float32)

x = x.reshape(-1,3,1)
print(x.shape)

model = Sequential([
    Input(shape=(3,1)),
    SimpleRNN(8, activation= "relu"),
    Dense(1)
])

model.summary()

model.compile(
    loss = "mse",
    optimizer = "adam"
)

model.fit(
    x,
    y,
    epochs = 500,
    verbose=1
)
#  3일후 아래의 가격 다음알 을 예측하시오.

# 예측
x_pred = np.array([
    [[72500],
     [73000],
     [73500]]
])
x_pred = x_pred.reshape(-1,3,1)

pred = model.predict(x_pred)
print("다음 날 예상 종가",pred)
#    
```

```python
# Feature 1 : 온도(Temperature)
# Feature 2 : 습도(Humidity)

# 즉, 3일간의 온도와 습도를 보고 다음 날의 온도를 예측하는 예제

# [온도, 습도]
x = np.array([
    [[20,60],[21,61],[22,62]],
    [[23,63],[24,64],[25,65]],
    [[26,66],[27,67],[28,68]]
])

# 다음 날 온도
y = np.array([
    23,
    26,
    29
])

x = x.reshape(-1,3,2)
print(x.shape)

# ==========================================
# data_size = 3
# time_steps = 3
# features = 2
# ==========================================

model = Sequential([
    Input(shape=(3,2)),
    SimpleRNN(8,activation="relu"),
    Dense(1)
])

model.summary()

model.compile(
    loss = "mse",
    optimizer = "adam"
)

model.fit(
    x,
    y,
    epochs = 500,
    verbose=1
)
```

```python
# 예측하시오
#    [[29,69],
#      [30,70],
#      [31,71]]

# 예측
x_pred = np.array([
    [
        [29,69],
        [30,70],
        [31,71]
    ]
])
x_pred = x_pred.reshape(-1,3,2)

pred = model.predict(x_pred)
print("다음 날 예상온도",pred)
```

```python
# ==========================================
# 1. 데이터 준비
# ==========================================

# [키(cm), 몸무게(kg), 나이]
x = np.array([
    [[170,60,20], [171,61,21]],
    [[180,70,30], [181,71,31]],
    [[190,80,40], [191,81,41]],
    [[200,90,50], [201,91,51]]
])

# 다음 해 키
y = np.array([
    172,
    182,
    192,
    202
])

x = x.reshape(-1,2,3)
print(x.shape)

# ==========================================
# data_size = 4
# time_steps = 2
# features = 3
# ==========================================

model = Sequential([
    Input(shape=(2,3)),
    SimpleRNN(8,activation="relu"),
    Dense(1)
])

model.summary()

model.compile(
    loss = "mse",
    optimizer = "adam"
)

model.fit(
    x,
    y,
    epochs = 500,
    verbose=1
)

x_pred = np.array([
    [
        [210,95,25],
        [211,96,26]
    ]
])

x_pred = x_pred.reshape(-1,2,3)
pred = model.predict(x_pred)
print("다음 해 키",pred)
```

```python
# [키(cm), 몸무게(kg), 나이]
x = np.array([
    [[170,60,20], [171,61,21]],
    [[180,70,30], [181,71,31]],
    [[190,80,40], [191,81,41]],
    [[200,90,50], [201,91,51]]
])

# 다음 해 키, 몸무게
y = np.array([
    [172,62],
    [182,72],
    [192,82],
    [202,92]
])

x = x.reshape(-1,2,3)
print(x.shape)

model = Sequential([
    Input(shape=(2,3)),
    SimpleRNN(8,activation="relu"),
    Dense(2) # 출력이 2개
])

model.summary()

model.compile(
    loss = "mse",
    optimizer = "adam"
)

model.fit(
    x,
    y,
    epochs = 500,
    verbose=1
)

x_pred = np.array([
    [
        [175,65,25],
        [176,66,26]
    ]
])

x_pred = x_pred.reshape(-1,2,3)
pred = model.predict(x_pred)
print("다음 해 키",pred)
```
